```python
# Tourismusdaten Deutschland - Istrien einlesen 
df_tourism_germany = pd.read_csv(
    "data/raw/tourism_germany_istria_2019_2025.csv"
)

df_tourism_germany.head()
```

```python
# Anzahl der Zeilen und Spalten 
df_tourism_germany.shape
```

```python
# Spaltennamen der Datensatzes prüfen 
df_tourism_germany.columns
```

```python
# Tourismusdaten korrekt einlesen
df_tourism_germany = pd.read_csv(
    "data/raw/tourism_germany_istria_2019_2025.csv",
    skiprows=2
)

df_tourism_germany.head()
```

```python
# Datentypen und Struktur des Datensatzes prüfen 
df_tourism_germany.info()
```

```python
# Fehlende Werte prüfen
df_tourism_germany.isna().sum()
```

```python
# Duplikate prüfen 
df_tourism_germany.duplicated().sum()
```

```python
# Spaltennamen des Datensatzes prüfen 
df_tourism_germany.columns.tolist()
```

```python
# Toursimusdaten vom Wide Format ins Long Format umformen
df_tourism_germany_long = df_tourism_germany.melt(
    id_vars=["Spatial unit", "Countries"],
    var_name="Zeitraum_Kennzahl",
    value_name="Wert"
)

df_tourism_germany_long.head()
```

```python
df_tourism_germany_long.shape
```

```python
# Jahr aus der Spalte Zeitraum_Kennzahl extrahieren
df_tourism_germany_long["Jahr"] = (
    df_tourism_germany_long["Zeitraum_Kennzahl"].str[:4]
)

df_tourism_germany_long.head()
```

```python
# Monat aus der Spalte Zeitraum_Kennzahl extrahieren 
df_tourism_germany_long["Monat"] = (
    df_tourism_germany_long["Zeitraum_Kennzahl"].str[5:7]
)

df_tourism_germany_long.head()
```

```python
# Kennzahl aus der Spalte Zeitraum_Kennzahl extrahieren 
df_tourism_germany_long["Kennzahl"] = (
    df_tourism_germany_long["Zeitraum_Kennzahl"].str.extract(r"(Tourist arrivals|Tourist nights)")
)

df_tourism_germany_long.head()
```

```python
# Jahressumme im Datensatz prüfen
df_tourism_germany_long.loc[
    df_tourism_germany_long["Zeitraum_Kennzahl"].str.contains("01.-12.", regex=False)
]
```

```python
# Jahressummen in der Spalte Monat kennzeichnen
df_tourism_germany_long.loc[
    df_tourism_germany_long["Zeitraum_Kennzahl"].str.contains("01.-12.", regex=False),
    "Monat"
] = "Gesamt"

df_tourism_germany_long.tail()
```

```python
# Kennzeichnung der Jahressumen prüfen 
df_tourism_germany_long["Monat"].value_counts()
```

```python
# Kennzahlen in separate Spalte umformen 
df_tourism_germany_clean = df_tourism_germany_long.pivot(
    index=["Spatial unit", "Countries", "Jahr", "Monat"],
    columns="Kennzahl",
    values="Wert"
).reset_index()

df_tourism_germany_clean.head()
```

```python
# Namen der Spaltenachse entfernen
df_tourism_germany_clean.columns.name = None

df_tourism_germany_clean.head()
```

```python
# Struktur der bereinigten Datensatzes prüfen 
df_tourism_germany_clean.shape
```

```python
df_tourism_germany_clean.info()
```

```python
# Datentype der Spalte Jahr anpassen
df_tourism_germany_clean["Jahr"] = (
    df_tourism_germany_clean["Jahr"].astype("int64")
)

df_tourism_germany_clean.info()
```

```python
# Fehlende Werte erneut im bereinigten Datensatz prüfen 
df_tourism_germany_clean.isna().sum()
```

```python
# Bereinigte Tourismusdaten Deutschland - Istrien speichern 
df_tourism_germany_clean.to_csv(
    "data/processed/tourism_germany_istria_clean.csv",
    index=False
)
```
